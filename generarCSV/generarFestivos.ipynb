```python
import pandas as pd
import holidays
from datetime import datetime

# Rango desde 2013 hasta 2026
fechas = pd.date_range(start='2013-01-01', end=datetime.now())
festivos_es = holidays.Spain(years=range(2013, 2027))

datos = []
for fecha in fechas:
    datos.append({
        "fecha": fecha.strftime("%Y-%m-%d"),
        "es_festivo": fecha in festivos_es
    })

df_festivos = pd.DataFrame(datos)
# Guardamos directamente en la capa bronce
df_festivos.to_csv("../data_lake/bronce/festivos_2013_actualidad.csv", index=False)
print("¡Tercera fuente generada y guardada en la capa Bronce!")
```
