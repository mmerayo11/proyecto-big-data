```python
import os
import time
import requests
import pandas as pd

print("Iniciando extracción RAW desde la API Pública de Red Eléctrica (REData)...")

# 1. Asegurarnos de que la carpeta de la Capa Bronce existe
ruta_bronce = "../data_lake/bronce"
os.makedirs(ruta_bronce, exist_ok=True)

# 2. Rango de años a extraer
años = range(2013, 2027)
lista_datos = []

# 3. Bucle de Ingesta (Solo lectura, sin transformación)
for año in años:
    print(f"Descargando demanda del año {año}...")
    
    url = f"https://apidatos.ree.es/es/datos/demanda/evolucion?start_date={año}-01-01T00:00&end_date={año}-12-31T23:59&time_trunc=day"
    
    respuesta = requests.get(url)
    
    if respuesta.status_code == 200:
        datos_json = respuesta.json()
        
        try:
            # Extraemos el nodo 'values' del JSON tal cual viene
            valores = datos_json['included'][0]['attributes']['values']
            lista_datos.extend(valores)
        except (KeyError, IndexError):
            print(f"Aviso: Estructura inesperada en el JSON del año {año}")
            
    else:
        print(f"Error al descargar {año}. Código HTTP: {respuesta.status_code}")
    
    time.sleep(1)

df_raw = pd.DataFrame(lista_datos)

ruta_salida_csv = f"{ruta_bronce}/demanda_ESIOS_raw.csv"
df_raw.to_csv(ruta_salida_csv, index=False, encoding='utf-8')

print(f"Archivo RAW generado: {ruta_salida_csv}")
```
