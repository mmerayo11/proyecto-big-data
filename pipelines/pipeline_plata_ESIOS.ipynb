```python
import pandas as pd

print("Leyendo el CSV de demanda")
df_demanda = pd.read_csv("../data_lake/bronce/demanda_ESIOS_raw.csv")

# Limpieza básica (quedarnos solo con lo útil para la demanda eléctrica)
df_demanda['fecha'] = pd.to_datetime(df_demanda['datetime'], utc=True).dt.date
df_demanda.rename(columns={'value': 'demanda_mwh'}, inplace=True)
columnas_utiles = ['fecha', 'demanda_mwh']
columnas_presentes = [col for col in columnas_utiles if col in df_demanda.columns]
df_limpio = df_demanda[columnas_presentes].copy()

# Normalizar tipos de datos (Convertir la columna fecha a formato datetime y demanda a número)
df_limpio['fecha'] = pd.to_datetime(df_limpio['fecha'], errors='coerce')
df_limpio['demanda_mwh'] = pd.to_numeric(df_limpio['demanda_mwh'], errors='coerce')

# Eliminar posibles duplicados de fechas por seguridad y ordenar cronológicamente
df_limpio = df_limpio.drop_duplicates(subset=['fecha']).sort_values('fecha').reset_index(drop=True)

# Guardar en la capa Plata en formato Parquet
ruta_plata = "../data_lake/plata/demanda_limpia.parquet"
df_limpio.to_parquet(ruta_plata, index=False, engine='pyarrow')

print(f"¡Éxito! Datos limpiados y guardados en: {ruta_plata}")
```
