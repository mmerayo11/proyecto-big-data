```python
import pandas as pd

print("Transformando Festivos a formato Parquet...")
# 1. Leer de Bronce
df_festivos = pd.read_csv("../data_lake/bronce/festivos_2013_actualidad.csv")

# 2. Asegurarnos de que la columna fecha sea un objeto de tiempo real (crucial para cruzar después)
df_festivos['fecha'] = pd.to_datetime(df_festivos['fecha'])

# 3. Guardar en Plata
ruta_festivos = "../data_lake/plata/festivos_limpios.parquet"
df_festivos.to_parquet(ruta_festivos, index=False, engine='pyarrow')

print(f"¡Capa Plata completada para los Festivos! Archivo en: {ruta_festivos}")
```
