```python
import pandas as pd

print("Leyendo el CSV")
# 1. Leer de la capa Bronce
df_clima = pd.read_csv("../data_lake/bronce/aemet_todas_estaciones_2013_actualidad.csv")

# 2. Limpieza básica (quedarnos solo con lo útil para la demanda eléctrica)
columnas_utiles = ['fecha', 'provincia', 'tmax', 'tmin', 'tmed']

# Nos aseguramos de que existan esas columnas antes de filtrar
columnas_presentes = [col for col in columnas_utiles if col in df_clima.columns]
df_limpio = df_clima[columnas_presentes].copy()

# 3. Normalizar tipos de datos (Convertir la columna fecha a formato datetime)
df_limpio['fecha'] = pd.to_datetime(df_limpio['fecha'], errors='coerce')

# 4. Guardar en la capa Plata en formato Parquet
ruta_plata = "../data_lake/plata/aemet_limpio.parquet"
df_limpio.to_parquet(ruta_plata, index=False, engine='pyarrow')

print(f"¡Éxito! Datos limpiados y guardados en: {ruta_plata}")
```
