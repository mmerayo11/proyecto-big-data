```python
import boto3
import os
from dotenv import load_dotenv
from botocore.exceptions import NoCredentialsError, ClientError

load_dotenv()

# Configuración
ACCESS_KEY = os.getenv('AWS_ACCESS_KEY_ID')
SECRET_KEY = os.getenv('AWS_SECRET_ACCESS_KEY')
SESSION_TOKEN = os.getenv('AWS_SESSION_TOKEN')
REGION = os.getenv('AWS_DEFAULT_REGION')
BUCKET_NAME = os.getenv('AWS_BUCKET_NAME')

# Archivos locales que queremos subir
archivos_a_subir = [
    {"local": "../data_lake/bronce/aemet_todas_estaciones_2013_actualidad.csv", "s3_key": "clima/aemet_historico.csv"},
    {"local": "../data_lake/bronce/festivos_2013_actualidad.csv", "s3_key": "contexto/festivos.csv"},
    {"local": "../data_lake/bronce/demanda_ESIOS_raw.csv", "s3_key": "electrica/demanda.csv"}
]

def subir_archivos_a_s3():
    print(f"Iniciando conexión con AWS S3 (Bucket: {BUCKET_NAME})...")
    
    try:
        # Crear el cliente de S3
        s3_client = boto3.client(
            's3',
            aws_access_key_id=ACCESS_KEY,
            aws_secret_access_key=SECRET_KEY,
            aws_session_token=SESSION_TOKEN,
            region_name=REGION
        )
        
        # Bucle para subir cada archivo
        for archivo in archivos_a_subir:
            ruta_local = archivo['local']
            ruta_s3 = archivo['s3_key']
            
            if os.path.exists(ruta_local):
                print(f"Subiendo {ruta_local} a S3 ({ruta_s3})...")
                s3_client.upload_file(ruta_local, BUCKET_NAME, ruta_s3)
                print(f"¡Subido con éxito!")
            else:
                print(f"Aviso: No se encontró el archivo local {ruta_local}")
                
        print("Proceso de sincronización con la nube finalizado.")

    except NoCredentialsError:
        print("Error: No se encontraron las credenciales de AWS. Revisa el archivo .env")
    except ClientError as e:
        print(f"Error de AWS: {e}")
    except Exception as e:
        print(f"Error inesperado: {e}")

# Ejecutar la función
subir_archivos_a_s3()
```
